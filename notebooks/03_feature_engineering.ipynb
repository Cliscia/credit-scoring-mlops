## Hipóteses de Feature Engineering

As features candidatas foram definidas a partir de hipóteses relacionadas ao
problema de risco de crédito.

| Feature candidata | Hipótese |
|---|---|
| `monthly_credit_burden` | O valor do crédito relativo ao prazo pode representar melhor a pressão financeira do empréstimo do que valor e duração isoladamente. |
| `age_group` | A relação entre idade e risco pode ser não linear, tornando faixas etárias potencialmente mais informativas. |
| `long_duration` | Operações de crédito com prazos elevados podem apresentar comportamento de risco diferente. |
| `high_credit_amount` | Valores elevados de crédito podem representar maior exposição financeira. |
| `financial_capacity` | A combinação de informações sobre conta corrente, poupança e emprego pode representar melhor a capacidade financeira do cliente. |

As hipóteses serão investigadas antes da implementação definitiva das features.
Features sem justificativa ou ganho mensurável poderão ser descartadas.

```python
# Cria uma feature experimental representando o valor do crédito
# dividido pela duração da operação em meses.

df_experiment = df.copy()

df_experiment["credit_amount_per_month"] = (
    df_experiment["credit_amount"] / df_experiment["duration_months"]
)

df_experiment[
    [
        "credit_amount",
        "duration_months",
        "credit_amount_per_month",
        "credit_risk",
    ]
].head()
```
